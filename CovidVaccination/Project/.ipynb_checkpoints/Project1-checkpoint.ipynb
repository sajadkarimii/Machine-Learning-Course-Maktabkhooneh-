## Importing the libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## Load and Prepare Data

```python
data = pd.read_csv("country_vaccinations_by_manufacturer.csv")
```

## Data Preprocessing

```python
data.isna().sum()
```

```python
data.date = pd.to_datetime(data.date)
```

```python
data["year"] = data["date"].dt.year
data["month"] = data["date"].dt.month
data["day"] = data["date"].dt.day
data["day_of_week"] = data["date"].dt.day_of_week
data["day_of_year"] = data["date"].dt.day_of_year
```

```python
data
```

## Strorytelling - Visualization

```python
(total/ total.sum() * 100).reset_index().plot(x="vaccine",kind="bar")
```

```python
data[["year","vaccine","total_vaccinations"]].groupby(["year","vaccine"]).max().groupby("year").sum().plot(kind="bar")
```

## Prepare Data for Machine learning

```python
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.model_selection import train_test_split
```

```python
x = data.drop(columns=["total_vaccinations","date"])
```

```python
y = data["total_vaccinations"]
```

```python
encoder = LabelEncoder()
```

```python
for col in ["vaccine","location"]:
    encoder = LabelEncoder()
    x[col] = encoder.fit_transform(x[col])
```

```python
scaler = StandardScaler()
```

```python
x = scaler.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.1)
```

## Train your model

```python
from xgboost import XGBRFRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import root_mean_squared_error
```

```python
model = XGBRFRegressor()
```

```python
param = {"learning_rate":[1,0.1,0.01],
        "max_depth":[None,5,10],
        "n_estimators":[100,300,500]}
```

```python
gsv = GridSearchCV(model,param).fit(x_train,y_train)
```

```python
gsv.best_params_
```

```python
gsv.best_score_
```

```python
best_model = gsv.best_estimator_
```

```python
from pycaret.regression import *
```

```python
reg = setup(data=data,target="total_vaccinations",categorical_features=["location", "vaccine"],numeric_features=["year","month","day","day_of_week","day_of_year"],)
```

```python
best_model = compare_models()
```

## Test the model and show the metrics

```python
best_model.score(x_test,y_test)
```

```python
pred = best_model.predict(x_test)
```

```python
root_mean_squared_error(y_test,pred)
```

## Save your final model

```python
from pycaret.regression import save_model
```

```python
save_model(best_model, 'my_best_model')
```
